```python
import pandas as pd
import requests
import time
import os
from tqdm import tqdm
```

```python
df_krut = pd.read_csv('learner_corpus.csv', sep=';;;')
df_krut
```

```python
output_dir = 'ru_standard'
os.makedirs(output_dir, exist_ok=True)
for index, text in tqdm(df_krut['body'].items()):
    if text.strip and text.strip() != 'loaded from xml':
        with open(f'{output_dir}/text_{index}.txt', 'w') as f:
            f.write(text)
```

```python
headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "*/*",
    "Connection": "keep-alive",
}

def download_file(i, save_dir):
    url = f"http://www.web-corpora.net/RLC/download_file/{i}/text"
    
    try:
        r = requests.get(url, headers=headers, timeout=10)
        
        if r.status_code == 200 and r.content:
            filepath = os.path.join(save_dir, f"text_{i}.txt")
            with open(filepath, "wb") as f:
                f.write(r.content)
            # print(f"OK {i}")
        else:
            print(f"FAIL {i} status={r.status_code}")
    
    except Exception as e:
        print(f"ERROR {i}: {e}")


save_dir = "ru_non_standard"
os.makedirs(save_dir, exist_ok=True)

for i in tqdm(range(3000, 3283)):
    download_file(i, save_dir)
    time.sleep(1)
```
